```python
import os
from getpass import getpass

os.environ["GROQ_API_KEY"] = getpass("Enter your Groq API key: ")
```

```python
%pip install -q -U langchain-text-splitters
from youtube_transcript_api import YouTubeTranscriptApi, TranscriptsDisabled
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_groq import ChatGroq
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_community.vectorstores import FAISS
from langchain_core.prompts import PromptTemplate
```

```python

video_id = "YbY8cVwWAvw"

try:
    transcript_data = YouTubeTranscriptApi().fetch(
        video_id,
        languages=["en"]
    )

    transcript = " ".join(
        snippet.text for snippet in transcript_data
    )

    print(transcript)

except TranscriptsDisabled:
    print("No captions available for this video.")
```

```python
transcript_data
```

**NOW WE WILL WE DOING THE INDEXING OF THE CHUNKS**

```python
splitter = RecursiveCharacterTextSplitter(chunk_size=1000, chunk_overlap=200)
chunks = splitter.create_documents([transcript])
```

```python
len(chunks)
```

```python
chunks[10]
```

**NOW EMBEDDING GENERATION AND STORING IT IN A VECTOR STORE
**

```python
%pip install -q -U langchain-huggingface sentence-transformers faiss-cpu
```

```python
from langchain_huggingface import HuggingFaceEmbeddings

embeddings = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)
```

```python

vector_store = FAISS.from_documents(chunks, embeddings)
```

```python
vector_store.index_to_docstore_id
```

```python
vector_store.get_by_ids(['1a81a9bc-4f63-43ef-8ed3-a3d49fa61c7d'])
```

**STEP 2 RETRIVAL**


```python
retriever = vector_store.as_retriever(search_type="similarity", search_kwargs={"k": 4})
```

```python
retriever
```

```python
retriever.invoke("What is shortest path")
```

**STEP 3 AUGUMENTATION**

```python
llm = ChatGroq(
    model="openai/gpt-oss-20b",
    temperature=0.2
)
```

```python
prompt = PromptTemplate(
    template="""
      You are a question-answering assistant for a video transcript.

Answer the question using ONLY the information in the context below.

Rules:
1. If the context does not contain the answer, say:
   "I don't know based on this video's transcript."
2. Do not use your general knowledge to fill missing information.
3. Do not invent facts, explanations, examples, or quotes.
4. If the question is unrelated to the video, say you don't know.
5. If the context provides only partial information, clearly state
   that the transcript does not provide enough information.
6. The question's wording alone is not evidence that its premise is true.

Context:
{context}

Question:
{question}

Answer:
    """,
    input_variables = ['context', 'question']
)
```

```python
question          = "what is the optimal solution of this code"
retrieved_docs    = retriever.invoke(question)
retrieved_docs
```

```python
context_text = "\n\n".join(doc.page_content for doc in retrieved_docs)
context_text
```

```python
final_prompt = prompt.invoke({"context": context_text, "question": question})
```

```python
final_prompt
```

**STEP 4 GENERATION**

```python
answer = llm.invoke(final_prompt)
print(answer.content)
```

**STEP 5 BUILDING A CHAIN**

```python
from langchain_core.runnables import RunnableParallel, RunnablePassthrough, RunnableLambda
from langchain_core.output_parsers import StrOutputParser
```

```python
def format_docs(retrieved_docs):
  context_text = "\n\n".join(doc.page_content for doc in retrieved_docs)
  return context_text
```

```python
parallel_chain = RunnableParallel({
    'context': retriever | RunnableLambda(format_docs),
    'question': RunnablePassthrough()
})
```

```python
parallel_chain.invoke('what is an edge in a graph')
```

```python
parser=StrOutputParser()
```

```python
main_chain = parallel_chain | prompt | llm | parser
```

```python
main_chain.invoke('Can you summarize the video')
```
